```python
def rev(s):
    i = len(s) -1
    s2 = ''
    while i >= 0:
        s2 += s[i]
        i -= 1
    return s2

rev_str3 = rev("Aaj wo bi paraye hue hai!")
rev(rev_str3)
```

```python
# Test the function
original_string = "hello"
reversed_string = rev(original_string)
print(reversed_string) # Output: "olleh"
```

```python
def rev2(s):
    i = len(s) - 1  # 1. Initialize the index 'i' to the last character
    s2 = ''
    while i >= 0:
        s2 += s[i]   # 2. Add the character at index 'i' to s2
        i -= 1       # 3. Decrement i by 1 (move one index to the left)
    return s2

# Test the function
original_string = "hello"
reversed_string = rev2(original_string)
print(reversed_string) # Output: "olleh"
```
